```python
'''

### 시험점수 관리하기

-학생들이 국어, 영어, 수학, 사회, 과학의 5 과목에 대해서 시험을 치렀습니다. 
-학생의 이름과 각 과목의 성적을 입력 받아 관리하는 프로그램을 만들어봅시다.

### 조건

- 국어, 영어, 수학, 사회, 과학 점수를 사용자에게 입력 받습니다.
- 최저 점수의 과목과 최고 점수의 과목, 그리고 5 과목의 평균 점수를 출력합니다.
    
'''
```

```python
# 입력 받기
name = input("학생 이름을 입력하세요: ")

subjects = ["국어", "영어", "수학", "사회", "과학"]
scores = {}

for subject in subjects:
    score = int(input(f"{subject} 점수를 입력하세요: "))
    scores[subject] = score

# 최고점과 최저점 과목 구하기
max_subject = max(scores, key=scores.get)
min_subject = min(scores, key=scores.get)

# 평균 계산
average_score = sum(scores.values()) / len(scores)

# 결과 출력
print(f"학생의 성적 분석 결과:")
print(f"가장 높은 점수의 과목: {max_subject} ({scores[max_subject]}점)")
print(f"가장 낮은 점수의 과목: {min_subject} ({scores[min_subject]}점)")
print(f"5과목 평균 점수: {average_score:.2f}점")
```

```python
# 과목 리스트
subjects = ["국어", "영어", "수학", "사회", "과학"]

# 입력 받기
scores_input = input("성적을 입력하세요. (국어 영어 수학 사회 과학):\n>>> ")
scores_list = list(map(int, scores_input.split()))

# 과목별 점수 딕셔너리로 저장
scores = dict(zip(subjects, scores_list))

# 최고/최저 점수 과목 구하기
max_subject = max(scores, key=scores.get)
min_subject = min(scores, key=scores.get)

# 평균 계산
average_score = sum(scores.values()) / len(scores)

# 결과 출력
print(f"최저 점수: {min_subject}")
print(f"최고 점수: {max_subject}")
print(f"평균 점수: {average_score:.1f}")
```
